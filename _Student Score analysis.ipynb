# Data Cleaning 

```python
df.isnull().sum() # here there is no null value in math , reading , writing score so we don't have to fill null values 
```

```python
df.drop("Unnamed: 0" , axis = 1) # to delete the column unnamed: 0
```

```python
df["WklyStudyHours"] = df["WklyStudyHours"].str.replace("05-Oct", "5-10")
```

```python
df
```

```python
plt.figure(figsize = (6,4))
ax = sns.countplot(data = df , x = "Gender")
ax.bar_label(ax.containers[0]) # to label the exact count of male and female 
plt.show()
```

```python
gb = df.groupby("ParentEduc").agg({"MathScore":'mean',"ReadingScore":'mean',"WritingScore":'mean'})
```

```python
sns.heatmap(gb,annot = True)
plt.show()
```
